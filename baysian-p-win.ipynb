```python
def draw_p_win(wins, losses, priors=0.5, return_y_limit=False):
    import numpy as np
    import seaborn as sns
    import matplotlib.pyplot as plt
    from scipy.stats import beta as beta_dist

    # Define parameters
    alpha = wins + priors
    beta = losses + priors
    mu = alpha / (alpha + beta)

    # Generate x values
    x = np.linspace(0, 1, 500)

    # Compute the beta distribution's probability density function (PDF)
    pdf = beta_dist.pdf(x, alpha, beta)

    # Plot the beta distribution using seaborn
    #sns.set(style="whitegrid")
    plt.figure(figsize=(8, 5))
    graph = sns.lineplot(x=x, y=beta_dist.pdf(x, alpha, beta))#, label=f"Beta(α={alpha}, β={beta})")

    plt.axvline(mu, 0, linestyle='--', color="crimson")

    # Customize the plot
    #plt.title("Beta Distribution", fontsize=16)
    plt.xlabel("P(win)", fontsize=14)
    plt.ylabel("Density", fontsize=14)
    #plt.legend(fontsize=12)
    if return_y_limit:
        y_min, y_max = graph.get_ylim()
        return y_max
    return plt
```

```python
graph = draw_p_win(2,1)
graph.show()
```

```python
graph = draw_p_win(6,4)
graph.show()
```

```python
def beta_sd(alpha, beta):
    return pow((alpha * beta) / (pow(alpha + beta, 2) * (alpha + beta + 1)), 0.5)

print(beta_sd(2.5, 1.5))
```
