```python
df = pd.read_csv("data/testingData.csv", low_memory=False)
teamsOrigin = pd.read_csv("the-blue-alliance-data/teams/teams.csv", low_memory=False)
teamsOrigin.to_dict("list")
```

```python
def win_probability(team1, team2):
    delta_mu = sum(r.mu for r in team1) - sum(r.mu for r in team2)
    sum_sigma = sum(r.sigma ** 2 for r in itertools.chain(team1, team2))
    size = len(team1) + len(team2)
    denom = math.sqrt(size * (BETA * BETA) + sum_sigma)
    ts = trueskill.global_env()
    return ts.cdf(delta_mu / denom)
```

```python
def getNumMatches(a = dict):
    matchNum = 0
    for x in a:
        matchNum = x
    return matchNum
```

# Matches Into Useable Format

```python
matchTeams = df[['team number', 'match number',  'win/lose', 'alliance position']].dropna()
matchTeams = pd.concat([matchTeams, df[['team number', 'match number', 'win/lose', 'alliance position']]])
matchTeamsD = matchTeams.to_dict('list')
matchNum = getNumMatches(matchTeams['match number'])
matches = {} #{match => [[r1, r2, r3],[b1, b2, b3], redWinLose]}
print(matchTeams)
matchNum += 1
for match in range(matchNum):
    if match == 0:
        continue
    redTeams = []
    blueTeams = []
    for x in range(3):
        num = matchTeamsD['team number'][match*6-6+x]
        str1 = f"{num}"
        redTeams.append(str1)
    for y in range(3):
        num = matchTeamsD['team number'][match*6-6+y+3]
        str1 = f"{num}"
        blueTeams.append(str1)
    redWin = matchTeamsD['win/lose'][match*6-6]
    matches[match] = [redTeams, blueTeams, redWin]
print(matches)
```

# Teams Initialized into TrueSkill

```python
env = trueskill.TrueSkill(draw_probability=.018)

teams = {}
for team in teamsOrigin['Key']:
    str = team
    teamSkill = env.create_rating()
    teams[team[3:None]] = [teamSkill, False]
print(teams)
```

```python
for match in range(matchNum):
    if match == 0:
        continue
    ratingTeams = [
        (teams[matches[match][0][0]][0], teams[matches[match][0][1]][0], teams[matches[match][0][0]][0],),
        (teams[matches[match][1][0]][0], teams[matches[match][1][1]][0], teams[matches[match][1][0]][0],)
    ]
    ranks = []
    if matches[1][2] == True:
        ranks = [0,1]
    else:
        ranks = [0,1]
    rated = env.rate(ratingTeams, ranks)
    for r in range(3):
        teams[matches[match][0][r]][0] = rated[0][r]
    for b in range(3):
        teams[matches[match][1][b]][0] = rated[1][r]


```
